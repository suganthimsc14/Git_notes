```python
# Histogram

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

insurance_df = pd.read_csv("Insurance.csv")

print(insurance_df.head())
.
print(insurance_df.info())

plt.figure(figsize=(8,5))

sns.histplot(insurance_df['age'], bins=20, kde=True, color='skyblue')

plt.title("Age Distribution")

plt.show()
```

```python
# Scatter Plot

plt.figure(figsize=(8,5))

sns.scatterplot(x='age', y='charges', hue='smoker', data=insurance_df )

plt.title("Charges vs Age (Smoker vs Non-Smoker)")

plt.show()
```

```python
# Box Plot)

plt.figure(figsize=(8,5))

sns.boxplot(x='region', y='charges', data=insurance_df)

plt.title("Charges Distribution by Region")

plt.show()
```

```python
# Bar Plot

plt.figure(figsize=(8,5))

sns.barplot(x='children', y='charges', data=insurance_df, estimator=sum)

plt.title("Total Charges by Number of Children")

plt.show()
```

```python
# Heatmap

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

insurance_df = pd.read_csv("Insurance.csv")

numeric_df = insurance_df.select_dtypes(include=['float64', 'int64'])

plt.figure(figsize=(8,6))

sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap (Numeric Features)")

plt.show()

```
